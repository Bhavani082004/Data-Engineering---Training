```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder \
    .appName("RetailTransactions") \
    .getOrCreate()

df = spark.read.csv(
    "retail_transactions_100k.csv",
    header=True,
    inferSchema=True
)

df.show(5)
df.printSchema()
print("Total records:", df.count())
```

**Exercise Set -1**

```python

df.show(20, truncate=False)


df.printSchema()

print(df.columns)


print(df.count())


df.select(
    "order_id",
    "product_name",
    "city",
    "quantity",
    "unit_price"
).show()


df.filter(F.col("category") == "Electronics").show()

df.filter(F.col("city") == "Hyderabad").show()


df.filter(F.col("quantity") > 3).show()


df.filter(F.col("unit_price") > 20000).show()

df.filter(F.col("status") == "Delivered").show()
```

**Exercise Set 2**

```python
df.filter(
    (F.col("category") == "Electronics") &
    (F.col("city") == "Hyderabad")
).show()

df.filter(
    (F.col("status") == "Delivered") &
    (F.col("region") == "South")
).show()

df.filter(
    (F.col("quantity") > 2) &
    (F.col("unit_price") > 10000)
).show()

df.filter(
    F.col("city").isin("Mumbai", "Pune")
).show()


df.filter(
    F.col("status") != "Cancelled"
).show()

df.filter(
    (F.col("category") == "Furniture") &
    (F.col("unit_price") > 8000)
).show()


df.filter(
    (F.col("category") == "Accessories") &
    (F.col("quantity") > 3)
).show()
```

**Exercise Set 3**

```python
df.select(
    "order_id",
    "customer_name",
    "product_name",
    "unit_price"
).show()


df_price = df.withColumnRenamed("unit_price", "price")
df_price.show(5)


df_customer = df.withColumnRenamed("customer_name", "customer")
df_customer.show(5)


df_no_rating = df.drop("rating")
df_no_rating.show(5)


df_dropped = df.drop(
    "customer_name",
    "rating",
    "payment_method"
)
df_dropped.show(5)
```

**Exercise Set 4**

```python
df_amounts = df.withColumn(
    "gross_amount",
    F.col("quantity") * F.col("unit_price")
)


df_amounts = df_amounts.withColumn(
    "discount_amount",
    F.col("gross_amount") * F.col("discount_pct") / 100
)


df_amounts = df_amounts.withColumn(
    "final_amount",
    F.col("gross_amount") - F.col("discount_amount")
)


df_amounts = df_amounts.withColumn(
    "country",
    F.lit("India")
)

df_amounts = df_amounts.withColumn(
    "high_value",
    F.when(F.col("final_amount") >= 50000, "Yes")
     .otherwise("No")
)


df_amounts.select(
    "order_id",
    "gross_amount",
    "discount_amount",
    "final_amount",
    "country",
    "high_value"
).show(10, truncate=False)
```

**Exercise Set 5**

```python
df_condition = df.withColumn(
    "quantity_category",
    F.when(F.col("quantity") >= 5, "Bulk")
     .otherwise("Regular")
)

df_condition = df_condition.withColumn(
    "price_category",
    F.when(F.col("unit_price") >= 10000, "High Price")
     .otherwise("Normal Price")
)


df_condition = df_condition.withColumn(
    "delivery_category",
    F.when(F.col("status") == "Delivered", "Completed")
     .otherwise("Not Completed")
)


df_condition.select(
    "order_id",
    "quantity",
    "unit_price",
    "status",
    "quantity_category",
    "price_category",
    "delivery_category"
).show(10, truncate=False)
```

**Exercise Set 11**

```python
df_dates = df.withColumn(
    "order_date",
    F.to_date(F.col("order_date"))
)

df_dates = df_dates.withColumn(
    "order_year",
    F.year(F.col("order_date"))
)

df_dates = df_dates.withColumn(
    "order_month",
    F.month(F.col("order_date"))
)

df_dates = df_dates.withColumn(
    "order_day",
    F.dayofmonth(F.col("order_date"))
)

df_dates = df_dates.withColumn(
    "day_of_week",
    F.date_format(F.col("order_date"), "EEEE")
)

df_dates.filter(
    F.year(F.col("order_date")) == 2025
).show()

df_dates.filter(
    F.month(F.col("order_date")) == 1
).show()

df_dates.select(
    "order_id",
    "order_date",
    "order_year",
    "order_month",
    "order_day",
    "day_of_week"
).show(10, truncate=False)
```

```python
from pyspark.sql import functions as F

df.select([
    F.count(
        F.when(F.col(c).isNull(), c)
    ).alias(c)
    for c in df.columns
]).show()


df_no_nulls = df.dropna()
print("Rows after removing missing values:", df_no_nulls.count())


df_no_missing_customer = df.filter(
    F.col("customer_name").isNotNull()
)
print("Rows with customer names:", df_no_missing_customer.count())


df_filled_rating = df.fillna(
    {"rating": 0}
)
df_filled_rating.select("rating").show(10)


df_filled_city = df.fillna(
    {"city": "Unknown"}
)
df_filled_city.select("city").show(10)


df_filled_quantity = df.fillna(
    {"quantity": 0}
)
df_filled_quantity.select("quantity").show(10)

df.select([
    F.count(
        F.when(F.col(c).isNull(), c)
    ).alias(c)
    for c in df.columns
]).show()
```
